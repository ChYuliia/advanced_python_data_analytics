````python
import pandas as pd
pop = pd.read_excel("population_usafacts.xlsx")
pop
````

Set an index on the year column.

````python
pop = pop.set_index('Year')
pop
````

## Draw some plots

Start with the default options.

````python
pop.reset_index().plot()
````

Well, that's a mess, isn't it?  Let's filter the data to get something more readable.

````python
pop[['Ohio','Indiana','Kentucky']].plot.line()
````

Let's work with more recent times.  Use a query to select years after 2000 and make a bar plot.

````python
pop.reset_index().query('Year > 2000').Ohio.plot.bar()
````

That's not very interesting-- the values are mostly the same for that time period.

Make a range of the years 1900, 1910, 1920, etc. and make a horizontal bar plot.

````python
pop.reset_index().query('Year % 10 == 0').Total.plot.barh()     
````

Make a density plot of Ohio's population.  This shows where most of the values are in the dataset

````python
pop.Ohio.plot.density()
````

Select the populations of Ohio, Indiana, and Kentucy for the year 2020, and draw a pie chart of the population distribution.

````python
pop.loc[2020, ['Ohio', 'Indiana', 'Kentucky']].plot.pie(
     title = "OKI Population DIstribution",
     ylabel = ''
)
````

Make side-by-side subplots of the Ohio, Indiana, and Kentucky populations.  Use the year as the *x* values, share the *y* axis, limit the years to 1990-2020.  Add titles, and rotate the text of the *x*-axis.

````python
pop[['Ohio', 'Indiana', 'Kentucky']].plot.line(
    title=['Ohio','Indiana','Kentucky'],
    xlabel='Year',
    ylabel='Population',
    sharey=True,
    rot = 45,
    xlim=(1990,2020),
    subplots=True,
    layout=(1,3)
)
````
