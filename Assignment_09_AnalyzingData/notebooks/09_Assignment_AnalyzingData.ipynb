## Your Dataset

```python
import pandas as pd
sales = pd.read_csv("sales.csv")
sales
```

*Independent Variables:*\
**Year** can take on 2 values: 2026 and 2027\
**Region** can take on 4 values: North, South, East and West\
**Type** can take on 2 values: Residential and Commercial\
**Product** can take on 4 values: Economy, Good, Better, Best\
2 times 4 times 2 times 4 = **64** individual data points.

*Value Variables:*\
Quantity_Sold\
Price_GBP\
Total_Sales_GBP

Quantity_Sold × Price_GBP = Total_Sales_GBP

With so many independent variables and value variables, groups, aggregates, and pivot tables give us easy ways to arrange the data along different axes.

Group the data by year and calculate the means of the value variables:

```python
sales.groupby("Year").mean()
```

Group the data by other independent variables.

```python
sales.groupby("Region").mean()
```

```python
sales.groupby("Product").mean()
```

Make a Pandas `DataFrameGroupBy` object, and apply aggregate functions.

```python
sales_group = sales.groupby(['Year','Region','Product']) \
    [['Transaction_ID','Quantity_Sold','Price_GBP','Total_Sales_GBP']]
sales_group.agg(['sum','mean','min','max'])
```

Make a pivot table, showing years for the rows, showing the regions in columns and quantity sold and price as the values.

```python
sales.pivot_table(
    index='Year', 
    columns='Region', 
    values=['Price_GBP','Quantity_Sold'], 
    aggfunc='mean'
)
```

Make several pivot table using different rows, columns, and values.

```python
sales.pivot_table(
    index='Region', 
    columns='Year',
    values=['Price_GBP','Quantity_Sold'], 
    aggfunc='mean'
)
```

```python
sales.pivot_table(
    index='Type',
    columns='Region',
    values='Price_GBP',
    aggfunc='mean'
)
```

```python
sales.pivot_table(
    index='Region',
    columns='Type',
    values='Total_Sales_GBP',
    aggfunc=['min','max']
)
```

```python
sales.pivot_table(
    index='Year',
    columns='Product', 
    values=['Price_GBP','Total_Sales_GBP'], 
    aggfunc=['min','max']
)
```

```python
sales.pivot_table(
    index=['Year','Product'],
    columns='Region',
    values=['Price_GBP','Quantity_Sold'],
    aggfunc='mean'
)
```

Calculate the percent change of total sales, grouped by year, then region.

```python
df = sales.pivot_table(
    index=['Year','Region'],
    values='Total_Sales_GBP',
    aggfunc='sum'
)
df.pct_change()
```
