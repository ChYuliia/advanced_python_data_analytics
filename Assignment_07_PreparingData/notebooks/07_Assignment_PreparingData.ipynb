# Dataset Summary
This dataset contains historical data on avocado prices and sales volume in multiple U.S. markets. 

One of the columns in this dataset, `Unnamed: 0`, contains sequential numbers that are irrelevant to analyzing this data. 

Three of the other columns contain sales for PLU (price look-up) codes `4046`, `4225`, and `4770`. These columns will also not be used by the analyses done in these projects.

If you review the data, you’ll see that some of the regions overlap. For example, one of the regions is the entire U.S., and all of the other regions are parts of the U.S. Because of that, you would need to review this data carefully before determining the best way to analyze it. For the purposes of these exercises, though, the overlapping regions won’t be taken into consideration.

```python
# Import required modules and libraries
import pandas as pd
```

```python
# Read the data from the CSV file into a DataFrame 
# and inspect the initial records
df_avocado = pd.read_csv("avocado.csv")
print("=> Initial five rows of the avocado dataset:\n")
display(df_avocado.head())

# Compute the number of missing values for each column 
# to ensure data integrity
print("\n=> Missing values summary by column:\n")
print(df_avocado.isnull().sum())

# Extract and inspect unique geographical regions 
print("\n=> Frequency count of unique regions:\n")
display(df_avocado.region.value_counts())

# Drop rows that have a region of TotalUS 
# and compute the number of dropped rows
total_count = df_avocado.shape[0]
df_avocado = df_avocado[df_avocado.region != 'TotalUS']
print("\n=> Total rows dropped after removing nationwide aggregates:", 
      total_count - df_avocado.shape[0])

# Drop the Unnamed: 0, 4046, 4225, and 4770 columns
# that are not required for analysis
columns_to_drop = ['Unnamed: 0', '4046', '4225', '4770']
df_avocado.drop(columns=columns_to_drop, inplace=True)

# Convert the Date column to the DateTime type.
df_avocado.Date = pd.to_datetime(df_avocado.Date)

# Standardize column headers by removing spaces and applying title case formatting
df_avocado.columns = df_avocado.columns.str.replace(" ","").str.strip()
df_avocado.columns = [col[0].upper() + col[1:] for col in df_avocado.columns]

# Calculate the proportion of extra-large bags relative to total bag volume
df_avocado['PctXLargeBags'] = (df_avocado.XLargeBags / df_avocado.TotalBags) * 100

# Calculate the proportion of large bags relative to total bag volume.
df_avocado['PctLargeBags'] = (df_avocado.LargeBags / df_avocado.TotalBags) * 100

# Calculate the proportion of small bags relative to total bag volume
df_avocado['PctSmallBags'] = (df_avocado.SmallBags / df_avocado.TotalBags) * 100

# Display the first five rows of data to view the new columns
print("\n=> Updated dataset featuring calculated bag size percentages:\n")
display(df_avocado.head())

# Extract a subset dataframe containing only the target dimensions for aggregation
df_avocado_new = df_avocado[['Region', 'Type', 'Year', 'TotalBags']].copy()

# Aggregate total bag volume by region, year, and type, 
# then pivot avocado types into separate columns
print("\n=> Final aggregated and unstacked summary table:\n")
df_avocado_new.groupby(['Region', 'Year', 'Type']).sum().unstack(level='Type')
```
