## 1.
Write a function to calculate the minimum, maximum, total, and average values of a list.  Return these four values as a tuple.

Display a menu like this:

```
1. Enter a value
2. Display statistics
3. Exit
```

Get user input and take the appropriate action.  Repeat until menu choice 3 is chosen.

Display statistics to one decimal place.  Beware of division by zero.

````python
def display_menu():
    print("1. Enter a value")
    print("2. Display statistics")
    print("3. Exit")

def stats(list1):    
    if len(list1) == 0:
        return (0,0,0,0)
        
    min_value = list1[0]
    max_value = list1[0]   
    total = 0
    
    for i in list1:
        total += i        
        if i < min_value:
            min_value = i
        if i > max_value:
            max_value = i

    avg = total / len(list1)
    return (min_value, max_value, total, avg)
    
list_values = []
choice = ""

while choice != "3":
    display_menu()
    choice = input("Choose an option: ")
    if (choice == "1"):
        input_value = input("Enter a value => ")
        list_values.append(float(input_value))    
    elif (choice == "2"):
        print("-"*25, "STATISTICS", "-"*25)
        mi, ma, t, a = stats(list_values)
        print("\n=> Minimum = {:.1f}; Maximum = {:.1f}; Total = {:.1f}; Average = {:.1f}\n".format(mi, ma, t, a))
        print("-"*62)
````
