## Data Usage

Increasingly, ours is an information society. Information (data) is a competitive advantage. Information is power.

__Ethical and Legal Restrictions on Using Data:__ Just because you can doesn’t mean you should.
- Data is private property, commonly known as Intellectual Property (IP)
- In the US, the creator of the data owns a copyright whether or not she registers it with the government
- Copyright law varies among countries
- Generally in USA, the default is to allow usage; in much of the rest of the world, especially the European Union (EU), the default is to prohibit. There are wide variations, so check your local jurisdiction.
- Using data created by others requires their permission
- Using data created by others requires proper attribution and citation
- Data has proprietary, confidentiality and competitive-advantage aspects. To make it exceedingly obvious, the UK calls its legislation "Trade-Secrets" act.
- Especially military documents and data can be classified as confidential, secret, top secret. You can do serious jailtime for violations.
- Your employer should have appropriate policies and procedures for using internally generated and external data
- Governments generally prohibit the use of personally-identifiable data, and have strict rules for creating personally identifiable data
- Likewise research institutions, such as universities, have "use of humans in research" policies and procedures
- Non-commercial usage, fair use, and creative commons regulations generally provide royalty-free use of data in teaching at universities where there is no profit component.
- Government-generated data is usually freely used for most purposes, as the data is in the public domain, being taxpayer funded. For example, US Census Bureau and Bureau of Labor Statistics data.
- Commercial use of external data usually involves royalties, user fees and subscriptions, with contractual terms and conditions for using, storing, modifying, combining and reselling resulting data products.
- State sponsored and private attackers may try to steal your IP.  How will you guard against it?
- LEGAL REQUIREMENTS FOR DATA USE AND STORAGE VARY SUBSTANTIALLY AMONG COUNTRIES. YOU MUST COMPLY WITH ALL LAWS AND REGULATIONS WHEREVER YOUR EMPLOYER HAS OPERATIONS AND DATA USAGE.

```python
import pandas as pd
from urllib import request
from zipfile import ZipFile
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the zip file to disk 
zip_url = 'https://www150.statcan.gc.ca/n1/tbl/csv/11100125-eng.zip'
request.urlretrieve(zip_url, filename='11100125-eng.zip')

# Extract the files and list their names 
file_names = list()
with ZipFile('11100125-eng.zip',mode='r') as zip:
    zip.extractall()
    for file in zip.infolist():
        file_names.append(file.filename)
        print(file.filename, file.compress_size, file.file_size)
data_file_name = [f for f in file_names if 'metadata' not in f.lower()][0]
```

```python
# Read it into a Python DataFrame
food_spending = pd.read_csv(data_file_name)
food_spending.head()
```

```python
# --- Exploratory Data Analysis (EDA) ---
print("=" * 50)
print(f"=> Dataset Dimensions (Rows, Columns): {food_spending.shape}")
print("=" * 50)

print("\n=> General Information on Data Types and Memory:")
food_spending.info()

print("\n" + "=" * 50)
print("=> Statistical Summary (Describe):")
print("=" * 50)
display(food_spending.describe())
```

```python
# Filter data for Canada and convert VALUE column to numeric type (coerce errors to NaN)
df_plot = food_spending[food_spending['GEO'] == 'Canada'].copy()
df_plot['VALUE'] = pd.to_numeric(df_plot['VALUE'], errors='coerce')

# Select specific summary-level categories to avoid cluttering the plot
top_categories = [
    'Food expenditures', 
    'Food purchased from stores', 
    'Food purchased from restaurants'
]
df_filtered = df_plot[df_plot['Food expenditures, summary-level categories'].isin(top_categories)]

# Initialize the matplotlib figure canvas
plt.figure(figsize=(12,6))

# Create the line plot using seaborn
sns.lineplot(data=df_filtered,
             x='REF_DATE', 
             y='VALUE', 
             hue='Food expenditures, summary-level categories',
             marker='o',
             linewidth=2.5)

# Customize plot titles, labels, legend, and grid styling
plt.title("Food Expenditures in Canada Over Time", 
          fontsize=16, fontweight='bold')
plt.xlabel("Year", fontsize=12)
plt.ylabel("Expenditure ($)",fontsize=12)
plt.legend(title="Category")
plt.grid(True, linestyle='--', alpha=0.3)
plt.tight_layout()

# Display the final visualization
plt.show()
```

```python
# Filter data for Canada and the latest available year
latest_year = food_spending['REF_DATE'].max()
df_snapshot = food_spending[
    (food_spending['GEO'] == 'Canada') & 
    (food_spending['REF_DATE'] == latest_year)
].copy()

# Convert VALUE to numeric and drop missing values
df_snapshot['VALUE'] = pd.to_numeric(df_snapshot['VALUE'], errors='coerce')
df_snapshot = df_snapshot.dropna(subset=['VALUE'])

df_snapshot = df_snapshot[df_snapshot['Food expenditures, summary-level categories'].isin(top_categories)]

# Create a clean horizontal bar plot for a single year
plt.figure(figsize=(10, 4))
sns.barplot(
    data=df_snapshot,
    y='Food expenditures, summary-level categories',
    x='VALUE',
    palette='Blues_r'
)

# Customizations
plt.title(f"Food Expenditures in Canada ({latest_year})", fontsize=14, fontweight='bold')
plt.xlabel("Expenditure ($)", fontsize=12)
plt.ylabel("Category", fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()

plt.show()
```
